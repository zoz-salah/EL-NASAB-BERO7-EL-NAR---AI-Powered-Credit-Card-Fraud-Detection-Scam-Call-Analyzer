# EL NASAB BERO7 EL NAR — EDA & Model Training

Download `creditcard.csv` from the Kaggle Credit Card Fraud Detection dataset and place it in `../backend/data/creditcard.csv` before running this notebook.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('../backend/data/creditcard.csv')
df.shape
```

```python
# Class imbalance check
counts = df['Class'].value_counts()
print(counts)
print(f"Fraud rate: {counts[1] / counts.sum() * 100:.3f}%")
counts.plot(kind='bar', title='Class distribution (0 = genuine, 1 = fraud)')
plt.show()
```

```python
# Amount distribution by class
df.groupby('Class')['Amount'].describe()
```

## Train and compare models

Reuses the training function from `backend/model.py` so the notebook and the API stay in sync.

```python
import sys
sys.path.append('../backend')
from model import train_and_compare

xgb_model, results, cm = train_and_compare(df)
results
```

```python
# Confusion matrix
import seaborn as sns
sns.heatmap(cm, annot=True, fmt='d', cmap='viridis')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('XGBoost Confusion Matrix')
plt.show()
```

```python
# Feature importance
importances = pd.Series(xgb_model.feature_importances_, index=df.drop(columns=['Class']).columns)
importances.sort_values(ascending=False).head(15).plot(kind='barh', title='Top 15 Feature Importances')
plt.gca().invert_yaxis()
plt.show()
```
